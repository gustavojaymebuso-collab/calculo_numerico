### Parte 1 – Implementação dos métodos

Implemente cada método em uma função independente no Python. Por exemplo:
- $bissecao(f, a, b, tol, max_{iter})$;
- $newton(f, df, x0, tol, max_{iter})$;
- $secante(f, x0, x1, tol, max_{iter})$.
Cada função deve retornar, no mínimo:
- a aproximação final da raiz;
- o número de iterações realizadas;
- uma lista com as aproximações calculadas em cada iteração;
- uma indicação de convergência ou de falha.
Use uma tolerância de $10^{−6}$

e um limite máximo de 100 iterações. O código deve detectar

situações em que não seja possível continuar, como:
- $f(a)$$f(b)$ > 0 no método da bisseção;
- derivada $$f′(x_{k})$$ igual ou muito próxima de zero no método de Newton;
- denominador igual ou muito próximo de zero no método da secante;
- ausência de convergência dentro do número máximo de iterações.
Para determinar a parada, você pode usar a diferença entre aproximações sucessivas,

$$|x_{k+1} − x_{k}| < tol,$$

e também verificar o resíduo,

$$|f(x_{k+1})| < tol.$$
Explique no notebook qual critério foi adotado.

Método da Bisseção

```python
def bissecao(f, a, b, tol = 1e-6, max_iter = 100):
  aproximacoes = []

  fa = f(a)
  fb = f(b)

  if fa * fb > 0:
    return None, 0, aproximacoes, "Não é possível continuar"

  for i in range(max_iter):
    m = (a + b) / 2
    fm = f(m)

    aproximacoes.append(m)
    if abs(fm) < tol or abs(b - a) < tol:
      return m, i + 1, aproximacoes, "Convergiu"

    if fa * fm > 0:
      a = m
      fa = fm
    else:
      b = m
      fb = fm

  return m, max_iter, aproximacoes, "Não convergiu"
```

Método de Newton

Usando a fórmula:
$$x_{k+1} = x_{k} - \frac{f(x_{k})}{f'(x_{k})}$$

```python
def newton(f, df, x0, tol = 1e-6, max_iter = 100):
  aproximacoes = [x0]

  x = x0

  for i in range(max_iter):
    derivada = df(x)
    if abs(derivada) < 1e-12:
      return x, i, aproximacoes, "Derivada muito próxima de zero"

    x_novo = x - f(x) / derivada

    aproximacoes.append(x_novo)

    if abs(x_novo - x) < tol:
      return x_novo, i + 1, aproximacoes, "Convergiu"

    x = x_novo

  return x, max_iter, aproximacoes, "Não convergiu"
```

Método da Secante

Usando a fórmula:
$$x_{k+1} = x_{k} - f(x_{k})\frac{x_{k} - x_{k-1}}{f(x_{k}) - f(x_{k-1})}$$

```python
def secante(f, x0, x1, tol = 1e-6, max_iter = 100):
  aproximacoes = [x0, x1]

  for i in range(max_iter):
    f0 = f(x0)
    f1 = f(x1)

    denominador = f1 - f0

    if abs(denominador) < 1e-12:
      return x1, i, aproximacoes, "Denominador muito próximo de zero"

    x2 = x1 - f1 * (x1 - x0) / denominador

    aproximacoes.append(x2)

    if abs(x2 - x1) < tol and abs(f(x2)) < tol:
      return x2, i + 1, aproximacoes, "Convergiu"

    x0 = x1
    x1 = x2

  return x1, max_iter, aproximacoes, "Não convergiu"
```

Definindo as funções

```python
def f1(x):
    return x**2 - 2

def df1(x):
    return 2*x

def f2(x):
    return x**3 - x - 2

def df2(x):
    return 3*x**2 - 1

def f3(x):
    return math.exp(-x) - x

def df3(x):
    return -math.exp(-x) - 1

def f4(x):
    return x**3 - 2*x + 2

def df4(x):
    return 3*x**2 - 2
```

### Parte 3 - Animação Gráfica

Bisseção

```python
def animar_bissecao(f, a, b, nome_arquivo="bissecao.gif"):
    raiz, it, aproximacoes, resultado = bissecao(f, a, b)

    intervalos = []

    atual_a = a
    atual_b = b

    fa = f(atual_a)

    for m in aproximacoes:
        fm = f(m)

        intervalos.append((atual_a, atual_b))

        if fa * fm > 0:
            atual_a = m
            fa = fm
        else:
            atual_b = m

    x = np.linspace(a - 0.5, b + 0.5, 500)
    y = [f(valor) for valor in x]

    fig, ax = plt.subplots(figsize=(8, 5))

    ax.plot(x, y, label="f(x)")
    ax.axhline(0, linestyle="--", label="y = 0")

    ax.set_xlabel("x")
    ax.set_ylabel("f(x)")
    ax.grid(True)

    def atualizar(frame):

        ax.clear()

        ax.plot(x, y, label="f(x)")
        ax.axhline(0, linestyle="--", label="y = 0")

        esquerda, direita = intervalos[frame]

        ax.plot(
            [esquerda, direita],
            [f(esquerda), f(direita)],
            marker="o",
            linestyle="None"
        )

        m = aproximacoes[frame]

        ax.plot(
            m,
            f(m),
            marker="o",
            markersize=8,
            label="Aproximação"
        )

        ax.axvline(esquerda, linestyle=":")
        ax.axvline(direita, linestyle=":")

        ax.set_title(
            f"Bisseção - Iteração {frame + 1}\n"
            f"x = {m:.8f} | |f(x)| = {abs(f(m)):.2e}"
        )

        ax.set_xlabel("x")
        ax.set_ylabel("f(x)")
        ax.grid(True)
        ax.legend()

        ax.set_xlim(a - 0.1, b + 0.1)

        ymin = min(y)
        ymax = max(y)
        ax.set_ylim(ymin, ymax)

    animacao = FuncAnimation(
        fig,
        atualizar,
        frames=len(aproximacoes),
        interval=700,
        repeat=False
    )

    animacao.save(
        nome_arquivo,
        writer=PillowWriter(fps=2)
    )

    plt.close()

    return animacao
```

```python
animar_bissecao(
    f1,
    1,
    2,
    "bissecao_f1.gif"
)
```

Newton

Fórmula reta tangente:
$$y = f(x_{k}) + f'(x_{k})(x - x_{k})$$

```python
def animar_newton(f, df, x0, nome_arquivo="newton.gif"):
    raiz, it, aproximacoes, resultado = newton(f, df, x0)

    x = np.linspace(x0 - 1, x0 + 1, 500)
    y = [f(valor) for valor in x]

    fig, ax = plt.subplots(figsize=(8, 5))

    def atualizar(frame):

        ax.clear()

        ax.plot(x, y, label="f(x)")
        ax.axhline(0, linestyle="--", label="y = 0")

        x_atual = aproximacoes[frame]
        y_atual = f(x_atual)

        ax.plot(
            x_atual,
            y_atual,
            marker="o",
            markersize=8,
            label="Aproximação"
        )

        derivada = df(x_atual)

        reta = y_atual + derivada * (x - x_atual)

        ax.plot(
            x,
            reta,
            linestyle="--",
            label="Reta tangente"
        )

        ax.set_title(
            f"Newton - Iteração {frame + 1}\n"
            f"x = {x_atual:.8f} | |f(x)| = {abs(y_atual):.2e}"
        )

        ax.set_xlabel("x")
        ax.set_ylabel("f(x)")
        ax.grid(True)
        ax.legend()

        ax.set_xlim(min(x), max(x))

    animacao = FuncAnimation(
        fig,
        atualizar,
        frames=len(aproximacoes),
        interval=700,
        repeat=False
    )

    animacao.save(
        nome_arquivo,
        writer=PillowWriter(fps=2)
    )

    plt.close()

    return animacao
```

```python
animar_newton(
    f1,
    df1,
    1.5,
    "newton_f1.gif"
)
```

Secante

```python
def animar_secante(f, x0, x1, nome_arquivo="secante.gif"):
    raiz, it, aproximacoes, resultado = secante(f, x0, x1)

    menor = min(aproximacoes) - 0.5
    maior = max(aproximacoes) + 0.5

    x = np.linspace(menor, maior, 500)
    y = [f(valor) for valor in x]

    fig, ax = plt.subplots(figsize=(8, 5))

    def atualizar(frame):

        ax.clear()

        # Função
        ax.plot(x, y, label="f(x)")
        ax.axhline(0, linestyle="--", label="y = 0")

        xa = aproximacoes[frame]
        xb = aproximacoes[frame + 1]

        ya = f(xa)
        yb = f(xb)

        ax.plot(
            [xa, xb],
            [ya, yb],
            marker="o",
            linestyle="None",
            markersize=8,
            label="Pontos"
        )

        if xb != xa:

            inclinacao = (yb - ya) / (xb - xa)

            reta = ya + inclinacao * (x - xa)

            ax.plot(
                x,
                reta,
                linestyle="--",
                label="Reta secante"
            )

        if frame + 2 < len(aproximacoes):

            proximo = aproximacoes[frame + 2]

            ax.plot(
                proximo,
                f(proximo),
                marker="o",
                markersize=8,
                label="Nova aproximação"
            )

        ax.set_title(
            f"Secante - Iteração {frame + 1}\n"
            f"x = {xb:.8f} | |f(x)| = {abs(yb):.2e}"
        )

        ax.set_xlabel("x")
        ax.set_ylabel("f(x)")
        ax.grid(True)
        ax.legend()

        ax.set_xlim(menor, maior)

    numero_frames = len(aproximacoes) - 2

    animacao = FuncAnimation(
        fig,
        atualizar,
        frames=numero_frames,
        interval=700,
        repeat=False
    )

    animacao.save(
        nome_arquivo,
        writer=PillowWriter(fps=2)
    )

    plt.close()

    return animacao
```

```python
animar_secante(
    f1,
    1,
    2,
    "secante_f1.gif"
)
```
